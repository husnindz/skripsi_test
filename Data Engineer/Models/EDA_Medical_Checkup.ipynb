# Exploratory Data Analysis — Medical Check-up Dataset

Notebook ini digunakan untuk melakukan **Exploratory Data Analysis (EDA)** pada dataset medical check-up.

Tahapan EDA:
1. Dataset Overview
2. Missing Value Analysis
3. Duplicate Analysis
4. Target Distribution
5. Feature Distribution
6. Categorical Feature Distribution
7. Outlier Analysis
8. Correlation Analysis
9. Feature Distribution by Target

**Catatan:** dataset yang digunakan adalah `merge_DocLab_balanced.csv`, yaitu dataset eksperimen yang telah diseimbangkan. Hasil distribusi target pada notebook ini tidak boleh dianggap sebagai distribusi dataset mentah/original.


## 1. Import Library & Load Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("merge_DocLab_balanced.csv")

print("Shape dataset :", df.shape)

print("\nNama kolom:")
print(df.columns.tolist())

print("\n5 data pertama:")
print(df.head(5))
```

## 2. Dataset Overview

```python
print("=== INFO DATASET ===")
df.info()

print("\n=== TIPE DATA ===")
print(df.dtypes)

print("\n=== STATISTIK DESKRIPTIF ===")
print(df.describe())
```

## 3. Missing Value Analysis

```python
print("=== MISSING VALUE ===")

missing = df.isnull().sum()

print(missing)
print("\nTotal missing value :", missing.sum())
```

## 4. Duplicate Analysis

```python
print("=== DUPLICATE DATA ===")

duplicate_count = df.duplicated().sum()

print("Jumlah duplicate :", duplicate_count)
```

## 5. Target Distribution

```python
target_labels = {
    0: "Sehat / Tidak Ada Indikasi",
    1: "Jantung dan Pembuluh Darah",
    2: "Penyakit Dalam",
    3: "Paru Paru"
}

target_counts = df["NM_UNIT"].value_counts().sort_index()

target_names = [
    target_labels[i]
    for i in target_counts.index
]

plt.figure(figsize=(10, 5))

bars = plt.bar(
    target_names,
    target_counts.values,
    edgecolor="black"
)

plt.title("Distribusi Target")
plt.xlabel("Kelas")
plt.ylabel("Jumlah Data")
plt.xticks(rotation=20)

for bar, count in zip(bars, target_counts.values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(count),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
target_percentage = (
    df["NM_UNIT"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

print("=== PERSENTASE DISTRIBUSI TARGET ===")

for class_id, percentage in target_percentage.items():
    print(
        f"{class_id} - "
        f"{target_labels[class_id]} : "
        f"{percentage:.2f}%"
    )
```

## 6. Feature Distribution

```python
numeric_features = [
    "UMUR_TAHUN",
    "cholesterol total",
    "creatinin",
    "fbs",
    "rbs",
    "hgb",
    "lymfosit%",
    "mch",
    "mchc",
    "mcv",
    "ureum",
    "wbc"
]

fig, axes = plt.subplots(
    4,
    3,
    figsize=(16, 16)
)

fig.suptitle(
    "Distribusi Fitur Medical Check-up",
    fontsize=16,
    fontweight="bold"
)

axes = axes.flatten()

for i, feature in enumerate(numeric_features):

    axes[i].hist(
        df[feature],
        bins=40,
        edgecolor="black",
        alpha=0.7
    )

    axes[i].set_title(feature)
    axes[i].set_xlabel("Value")
    axes[i].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## 7. Categorical Feature Distribution — JENIS_KELAMIN

```python
gender_counts = df["JENIS_KELAMIN"].value_counts().sort_index()

plt.figure(figsize=(7, 5))

bars = plt.bar(
    gender_counts.index.astype(str),
    gender_counts.values,
    edgecolor="black"
)

plt.title("Distribusi JENIS_KELAMIN")
plt.xlabel("JENIS_KELAMIN")
plt.ylabel("Jumlah Data")

for bar, count in zip(
    bars,
    gender_counts.values
):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(count),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

## 8. Outlier Analysis

```python
fig, axes = plt.subplots(
    4,
    3,
    figsize=(16, 14)
)

fig.suptitle(
    "Boxplot Fitur Medical Check-up",
    fontsize=16,
    fontweight="bold"
)

axes = axes.flatten()

for i, feature in enumerate(numeric_features):

    axes[i].boxplot(
        df[feature].dropna(),
        vert=True
    )

    axes[i].set_title(feature)
    axes[i].set_ylabel("Value")

plt.tight_layout()
plt.show()
```

### Catatan Interpretasi Outlier

Boxplot digunakan untuk **mengidentifikasi** nilai yang berada jauh dari distribusi utama. Nilai yang terdeteksi sebagai outlier tidak otomatis merupakan kesalahan data dan tidak langsung dihapus.

Pada data medical check-up, nilai ekstrem perlu ditinjau berdasarkan konteks data dan sumber pencatatannya sebelum menentukan tindakan preprocessing.


## 9. Correlation Analysis

```python
correlation = df[numeric_features].corr()

plt.figure(figsize=(12, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Matrix Fitur Medical Check-up"
)

plt.tight_layout()
plt.show()
```

### Catatan Interpretasi Korelasi

Correlation matrix digunakan untuk melihat hubungan linear antar fitur numerik.

Korelasi yang tinggi tidak otomatis menunjukkan hubungan sebab-akibat dan tidak otomatis berarti salah satu feature harus dihapus. Hasil ini digunakan sebagai informasi awal untuk memahami struktur dataset.


## 10. Feature Distribution by Target

```python
plot_features = [
    "UMUR_TAHUN",
    "cholesterol total",
    "creatinin",
    "fbs",
    "rbs",
    "hgb"
]

df_eda = df.copy()

df_eda["TARGET_NAME"] = df_eda["NM_UNIT"].map(
    target_labels
)

fig, axes = plt.subplots(
    2,
    3,
    figsize=(18, 10)
)

fig.suptitle(
    "Distribusi Fitur Berdasarkan Kelas Target",
    fontsize=16,
    fontweight="bold"
)

axes = axes.flatten()

for i, feature in enumerate(plot_features):

    sns.boxplot(
        data=df_eda,
        x="TARGET_NAME",
        y=feature,
        ax=axes[i]
    )

    axes[i].set_title(feature)
    axes[i].set_xlabel("Target")
    axes[i].set_ylabel(feature)

    axes[i].tick_params(
        axis="x",
        rotation=20
    )

plt.tight_layout()
plt.show()
```

## 11. EDA Summary

```python
eda_summary = pd.DataFrame({
    "dtype": df.dtypes,
    "missing": df.isnull().sum(),
    "unique": df.nunique(),
    "mean": df.mean(numeric_only=True),
    "std": df.std(numeric_only=True),
    "min": df.min(numeric_only=True),
    "max": df.max(numeric_only=True)
})

print(eda_summary)
```

## Kesimpulan EDA

Setelah menjalankan seluruh tahapan di atas, interpretasi EDA sebaiknya mencakup:

- kondisi struktur dataset,
- keberadaan atau tidaknya missing value,
- keberadaan atau tidaknya duplicate,
- distribusi target,
- bentuk distribusi masing-masing feature,
