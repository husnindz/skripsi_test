# Import Library & Data


```python
import os
import pandas as pd

# Path folder data lab (bisa dari folder root atau subfolder Lab/)
LAB_DIR = 'Lab' if os.path.exists('Lab') else '.'

def read_clean_csv(filepath):
    df = pd.read_csv(filepath, encoding='latin1')
    df.columns = [c.replace('\ufeff', '').replace('\xef\xbb\xbf', '').strip() for c in df.columns]
    return df
```

```python
# 1. Load Data Lab Februari
lab_feb = read_clean_csv(os.path.join(LAB_DIR, 'LAB_FEB.csv'))
```

```python
# 2. Load Data Lab Maret - April
lab_maret_april = read_clean_csv(os.path.join(LAB_DIR, 'MARET-APRIL06.csv'))
```

```python
lab_maret_april
```

```python
# 3. Load Data Lab Mei - Juni
lab_mei_juni = read_clean_csv(os.path.join(LAB_DIR, 'MEI_JUNI.csv'))
```

# Cleaning Data


```python
df_lab_all['NM_PERIKSA'].nunique()
```

```python
# Normalisasi nama parameter dan nama pemeriksaan
df_lab_all['NM_PARAM'] = df_lab_all['NM_PARAM'].str.strip().str.lower()
df_lab_all['NM_PERIKSA'] = df_lab_all['NM_PERIKSA'].str.strip().str.lower()
df_lab_all['TGL_INPUT'] = pd.to_datetime(df_lab_all['TGL_INPUT'])
```

```python
# Pivot table parameter laboratorium
lab_pivot_all = df_lab_all.pivot_table(
    index=['NO_REG', 'TGL_INPUT', 'NM_PERIKSA'],
    columns='NM_PARAM',
    values='HASIL',
    aggfunc='first'
).reset_index()
```

# Feature Selection


```python
target_tests = [
    'hematologi rutin',
    'creatinin',
    'ureum',
    'glukosa sewaktu',
    'hemoglobin'
]
```

```python
valid_patients = lab_pivot_all[lab_pivot_all['NM_PERIKSA'].isin(target_tests)]['NO_REG'].unique()
```

```python
lab_pivot_filtered = lab_pivot_all[lab_pivot_all['NO_REG'].isin(valid_patients)].copy()
```

```python
lab_pivot_filtered
```

```python
# Konversi nilai pemeriksaan menjadi numerik dan agregasi maksimum per pasien (NO_REG)
numeric_cols = lab_pivot_filtered.columns.drop(['NO_REG', 'TGL_INPUT', 'NM_PERIKSA'])
for col in numeric_cols:
    lab_pivot_filtered[col] = pd.to_numeric(lab_pivot_filtered[col], errors='coerce')

df_final_lab = lab_pivot_filtered.groupby('NO_REG').max().reset_index()
```
