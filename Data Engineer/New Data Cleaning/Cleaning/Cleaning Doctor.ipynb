# Import Library & Data


```python
import os
import pandas as pd

# Path folder data dokter (bisa dari folder root atau subfolder Dokter/)
DOKTER_DIR = 'Dokter' if os.path.exists('Dokter') else '.'

def read_clean_csv(filepath):
    df = pd.read_csv(filepath, encoding='latin1')
    df.columns = [c.replace('\ufeff', '').replace('\xef\xbb\xbf', '').strip() for c in df.columns]
    return df
```

```python
# 1. Load Data Dokter Februari
dokter_feb = read_clean_csv(os.path.join(DOKTER_DIR, 'dokter_feb.csv')).rename(columns={'TGL_LAHIR_F': 'TGL_LAHIR'})
```

```python
dokter_feb
```

```python
# 2. Load Data Dokter Maret - April
dokter_maret_april = read_clean_csv(os.path.join(DOKTER_DIR, 'DOKTER_MARET-APRIL06.csv'))
```

```python
dokter_maret_april
```

```python
# 3. Load Data Dokter Mei - Juni
dokter_mei_juni = read_clean_csv(os.path.join(DOKTER_DIR, 'DOKTER_MEI-JUNI06.csv'))
```

```python
dokter_mei_juni
```

## Feature Engineering(For Dashboard)


```python
dokter_feb['bulan'] = 'Februari'
dokter_maret_april['bulan'] = 'Maret-April'
dokter_mei_juni['bulan'] = 'Mei-Juni'
```

## Merge Table


```python
df_dokter_all = pd.concat(
    [dokter_feb, dokter_maret_april, dokter_mei_juni],
    ignore_index=True
)
```

# Cleaning Data


```python
df_dokter_all['NM_UNIT'].unique()
```

```python
selected_unit = [
    'SPES. PENYAKIT DALAM',
    'SPES. PARU PARU',
    'JANTUNG DAN PEMBULUH DARAH'
]

df_dokter_filtered = df_dokter_all[
    df_dokter_all['NM_UNIT'].isin(selected_unit)
]
```

```python
# Drop kolom identitas / tidak diperlukan
cols_to_drop = [c for c in ['NAMA', 'ASESMEN'] if c in df_dokter_filtered.columns]
if cols_to_drop:
    df_dokter_filtered = df_dokter_filtered.drop(columns=cols_to_drop)
```

```python
df_dokter_filtered.info()
```

Cek Jumlah Pasien yang periksa lebih dari 1 kali


```python
(df_dokter_filtered['NO_REG'].value_counts() > 1).sum()
```

Cek Jumlah Pasien yang periksa ke berbagai unit


```python
df_dokter_filtered.groupby('NO_REG')['NM_UNIT'].nunique().value_counts()
```

Hapus yang tidak ada diagnosa penyakit


```python
# Hapus baris tanpa diagnosa penyakit & validasi NO_REG
df_dokter_filtered = df_dokter_filtered[df_dokter_filtered['NM_PENYAKIT'].notna()]
df_dokter_filtered = df_dokter_filtered[df_dokter_filtered['NO_REG'].notna()]
df_dokter_filtered['NO_REG'] = df_dokter_filtered['NO_REG'].astype(int)
```

```python
# Pastikan 1 baris per pasien (NO_REG) agar rasio merge 1:1 sesuai format referensi
df_dokter_filtered = df_dokter_filtered.drop_duplicates(subset=['NO_REG'], keep='first')
df_dokter_filtered.info()
```

```python
df_dokter_filtered.to_csv('dokter_filtered.csv', index=False)
```
